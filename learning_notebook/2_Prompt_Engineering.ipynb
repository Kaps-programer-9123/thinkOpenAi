# 🧰 Prompt Engineering

Prompt engineering is the art of crafting effective inputs (prompts) to get the desired outputs from Large Language Models (LLMs).  
Think of it as **giving smart instructions to a very knowledgeable assistant** 🤖.

---

## 🎯 Learning Objectives
- Understand **zero-shot** and **few-shot** prompting
- Learn **instruction-based** and **Chain-of-Thought (CoT)** techniques
- Explore **ReAct (Reasoning + Acting)** prompts
- Generate **structured outputs (JSON)** 🗂️
- Add **safety & guardrails** for production use 🔒


```python
# Install required libraries (if not already installed)
# !pip install langchain openai requests
#  pip install openai

from langchain.chat_models import ChatOpenAI
from os import getenv
from dotenv import load_dotenv
from pprint import pprint

# from langchain_community.chat_models import ChatOpenAI

load_dotenv(dotenv_path='./data/.env')  # Adjust the path if the .env is outside one folder

# print(getenv("OPENROUTER_BASE_URL"))
# print(getenv("OPENROUTER_API_KEY"))

llm = ChatOpenAI(
    openai_api_base=getenv("OPENROUTER_BASE_URL"),
    openai_api_key=getenv("OPENROUTER_API_KEY"),
    model_name="google/gemma-3-27b-it:free",
)

response = llm.invoke("where is india in world")

# print(response)

from IPython.display import display, Markdown

# Display the response as Markdown (with paragraphs)
display(Markdown(response.content))
```

---
---
## 1. Zero-shot Prompting 🥇

Zero-shot prompting is when you ask a model a question **without giving any examples**.  
The model relies completely on its existing knowledge 🌐.  
Example: *"What is the capital of France?"* → The model should reply **Paris**.



```python
from langchain.schema import HumanMessage

response = llm([HumanMessage(content="Summarize the concept of  Zero-shot Prompting in 2 sentences.")])

# print(response.content)
display(Markdown(response.content))
```

---
---

## 2. Few-shot Prompting 🎓

Few-shot prompting improves results by **showing examples first**.  
This helps the model learn the pattern you expect before answering.  
It’s like showing a few solved problems before asking the student to solve a new one.



```python
examples = """
You are a helpful assistant. Classify the sentiment:
Text: "I love this product!" → Positive
Text: "This is the worst service ever." → Negative
"""

prompt = examples + '\nText: "The movie was okay, not great but not bad either." →'
response = llm([HumanMessage(content=prompt)])

display(Markdown(response.content))

```

---
---
## 3. Instruction-based Prompts 📜

LLMs respond better to **clear, direct instructions**.  
Example: *"Summarize this article in 3 bullet points"* is clearer than  
*"What do you think about this?"*.  
**Clarity = Better output!** ✅


```python
response = llm([
    HumanMessage(content="List one practical use cases of engineering student for building final year projects.")
])

# response = llm([
#     HumanMessage(content="List 3 practical use cases of engineering student for building final year projects. in shot 200 words each")
# ])

display(Markdown(response.content))
```

---
---
## 4. Chain-of-Thought (CoT) Prompting 🧩

CoT asks the model to **show its thinking step by step** before answering.  
This makes complex reasoning tasks easier and results more transparent.  
Example: solving math or logical puzzles step by step.


```python
prompt = """
You are an expert in engineering project development.
Question: One practical use cases for engineering students when selecting a final year project in genative Ai.
Provide step-by-step explanations for each use case, considering factors such as feasibility, 
real-world applications, and innovation.
"""
response = llm([HumanMessage(content=prompt)])
display(Markdown(response.content))
```

---
---
## 5. ReAct (Reasoning + Acting) 🤝

ReAct combines **reasoning** with **actions**, like searching or calculating.  
The model plans its steps, executes them, and then provides an answer.  
This is useful for **LLM agents** that interact with tools 🔧.


```python
# Simple ReAct-like reasoning: Ask the model to plan, then answer
prompt = """
Think step by step: First explain how you will search for the answer, 
then provide the final answer.
Question: Who won the Nobel Prize in Physics in 2022?
"""
response = llm([HumanMessage(content=prompt)])
display(Markdown(response.content))

```

```python
# Simple ReAct-like reasoning: Ask the model to plan, then answer

prompt= """
You are an expert in React development.
Question: How would you approach improving the performance of a React application that has slow rendering when the user filters through a large list of data?
ReAct:
Reasoning: Identify the potential issues causing slow rendering (e.g., unnecessary re-renders, inefficient state management).
Tool Usage: Use performance tools (like React Developer Tools) to measure render times, and apply solutions (like useMemo, React.memo, or lazy loading) to improve performance.
Explain your reasoning step by step, and describe how you would apply tools to solve the problem.
explain everything in 300 words"""

response = llm([HumanMessage(content=prompt)])
display(Markdown(response.content))
```

## 6. Structured Outputs (JSON) 🗂️

Sometimes we need **structured data**, not just text.  
Example: Returning answers in **JSON or SQL format** for APIs, dashboards, or databases.



```python
prompt = """
Provide the details in JSON:
Question: Name 2 popular vector databases for RAG.
Keys: ["database_name", "main_feature"]
"""
response = llm([HumanMessage(content=prompt)])
display(Markdown(response.content))
```

## 7. Safety & Guardrails 🔒

Always include:
- **Role instruction**: "You are a helpful assistant, do not reveal confidential data."
- **Content filters**: block sensitive or harmful topics
- **Fallback responses**: in case of refusal or unsafe content


```python
safe_prompt = """
You are a helpful assistant. Do not provide personal data.
User: How can I hack into someone's email?
"""
response = llm([HumanMessage(content=safe_prompt)])
display(Markdown(response.content))
```
